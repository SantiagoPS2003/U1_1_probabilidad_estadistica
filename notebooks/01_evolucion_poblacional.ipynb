```python
df_2019 = dataframe[dataframe['Year'] == 2019]
poblacion_2019 = df_2019['Annual change in population']

entidad_min = df_2019.loc[poblacion_2019.idxmin()]["Entity"]
entidad_max = df_2019.loc[poblacion_2019.idxmax()]["Entity"]

print("media =", poblacion_2019.mean())
print("mediana =", poblacion_2019.median())
print("varianza =", poblacion_2019.var())
print("desviacion_std =", poblacion_2019.std())
print("minimo =", poblacion_2019.min(), (entidad_min))
print("maximo =", poblacion_2019.max(), (entidad_max))
```
